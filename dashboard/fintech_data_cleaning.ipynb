```python
import pandas as pd

# Load the file directly from the Colab sidebar
df = pd.read_csv('digital_wallet_transactions.csv')

print("Total rows and columns:", df.shape)
df.head()
```

```python
# Step 1: Make all column names lowercase with underscores (easier for SQL)
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Step 2: Remove any duplicate transactions
df = df.drop_duplicates(subset=['transaction_id'])

# Step 3: Check if any columns have missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Step 1: Convert text date into a real date format
df['transaction_date'] = pd.to_datetime(df['transaction_date'])

# Step 2: Extract Month and Hour so we can spot trends later
df['txn_month'] = df['transaction_date'].dt.strftime('%Y-%m')
df['txn_hour'] = df['transaction_date'].dt.hour

# Step 3: Create a simple 1 or 0 column for Successful payments (like an Excel IF formula)
df['is_successful'] = (df['transaction_status'] == 'Successful').astype(int)

# Step 4: Calculate Net Revenue (Fee earned minus Cashback given)
df['net_revenue'] = df['transaction_fee'] - df['cashback']

# View the new columns to verify they worked
df[['transaction_id', 'transaction_status', 'is_successful', 'net_revenue', 'txn_month']].head()
```
